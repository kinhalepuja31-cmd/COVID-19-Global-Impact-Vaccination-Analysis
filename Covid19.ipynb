```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python


df = pd.read_csv("/content/complete covid-19.csv")
df.head()
```

```python
df.info()
```

Filter down to the essential indicators required for project

 Remove rows where continent is blank (these represent messy 'World' or    regional aggregates)

Set up a standard placeholder column to prevent any "44% Empty" errors in Power Query

Fill all early pandemic empty cells in vaccination columns with 0

Convert text dates into system datetime objects for smooth slicers in Power BI

Order the data chronologically by country and date

```python
keep_vars = [
    'date', 'country', 'continent',
    'total_cases', 'new_cases',
    'total_deaths', 'new_deaths',
    'total_vaccinations', 'people_vaccinated', 'people_fully_vaccinated'
]
df = df[keep_vars]



df = df.dropna(subset=['continent'])


df['Province/State'] = "Mainland"


vaccine_fields = ['total_vaccinations', 'people_vaccinated', 'people_fully_vaccinated']
df[vaccine_fields] = df[vaccine_fields].fillna(0)


for col in ['total_cases', 'new_cases', 'total_deaths', 'new_deaths']:
    df[col] = df[col].fillna(0)



df['Date'] = pd.to_datetime(df['date'])
df = df.drop(columns=['date'])


df = df.sort_values(by=['country', 'Date'])



df.to_csv("covid_cleaned_ready.csv", index=False)
print("🎉 Success! Your clean file is saved as: 'covid_cleaned_ready.csv'")
```
